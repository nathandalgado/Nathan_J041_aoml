# Stack Data Structure

## What is a Stack?

A **stack** is a linear data structure that follows the **Last In First Out (LIFO)** principle. The element inserted last is the first to be removed. Think of a stack of plates — you add to the top and remove from the top.

### Core Stack Operations:
- `push(x)`: Add an element `x` to the top of the stack.
- `pop()`: Remove and return the top element.
- `peek()`: View the top element without removing it.
- `isEmpty()`: Check if the stack is empty.
- `size()`: Return the number of elements in the stack.

Stacks are widely used in:
- Function call management (call stacks)
- Expression evaluation (postfix, infix)
- Undo operations in editors
- Depth-First Search (DFS)

---

## Pseudocode for Stack (Using Linked List)

```plaintext
class Node:
    data
    next

class Stack:
    top ← None
    size ← 0

    method push(data):
        node ← new Node(data)
        node.next ← top
        top ← node
        size ← size + 1

    method pop():
        if top is None:
            print "Stack is empty"
        else:
            data ← top.data
            top ← top.next
            size ← size - 1
            return data

    method peek():
        if top is not None:
            return top.data
        else:
            print "Stack is empty"


````python
class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None
        self.size = 0

    def push(self, data):
        node = Node(data)
        node.next = self.top
        self.top = node
        self.size += 1

    def pop(self):
        if self.top is None:
            print("Stack is empty")
            return None
        data = self.top.data
        self.top = self.top.next
        self.size -= 1
        return data

    def peek(self):
        if self.top is None:
            print("Stack is empty")
            return None
        return self.top.data

    def isEmpty(self):
        return self.size == 0

    def getSize(self):
        return self.size

# Example Usage
stack = Stack()
stack.push("egg")
stack.push("ham")
stack.push("spam")
print("Top element:", stack.peek())  # Output: spam
print("Popped:", stack.pop())        # Output: spam
print("Top after pop:", stack.peek())# Output: ham
````

# Asymptotic Growth: Prove Whether F(n) = O(G(n))

## Problem Statement

We are given:

- F(n) = log(n!)
- G(n) = n·log(n)

We are asked to verify whether:

F(n) = O(G(n))  
That is, determine if there exists a constant **c > 0** and **n₀ ≥ 0** such that:

log(n!) ≤ c · n·log(n) for all n ≥ n₀

---

## Step 1: Use Properties of Factorial

We know from logarithmic properties:

log(n!) = log(1 × 2 × 3 × ... × n)  
     = log(1) + log(2) + ... + log(n)

This is the sum of logarithms from 1 to n.

Now, observe that:

log(1) + log(2) + ... + log(n)  
 < ∫₁ⁿ log(x) dx + log(n)  
 = n·log(n) − n + 1  (by integration)

So:

log(n!) < n·log(n) − n + 1  
     < n·log(n) for all n ≥ 2

Thus, we can say:

log(n!) ≤ c · n·log(n)  
     with c = 1 and n₀ = 2

This satisfies the Big-O definition.

---

## Step 2: Alternative Approach Using Stirling’s Approximation

Stirling’s approximation provides an estimate of factorial:

n! ≈ sqrt(2πn) · (n/e)^n

Take logarithms:

log(n!) ≈ n·log(n) − n + 0.5·log(2πn)

Discarding lower-order terms:

log(n!) = Θ(n·log(n))

This means that log(n!) grows **asymptotically no faster than** n·log(n), and hence:

log(n!) = O(n·log(n))

---

## Step 3: Satisfying the Big-O Definition

By definition, F(n) = O(G(n)) means:

There exist constants **c > 0** and **n₀ ≥ 0** such that:

F(n) ≤ c · G(n) for all n ≥ n₀

We have shown:

log(n!) ≤ n·log(n) for n ≥ 2  
⇒ F(n) ≤ G(n)

Choose:
- c = 1
- n₀ = 2

Then the condition holds.

---

## Final Conclusion

Yes, the Big-O notation is satisfied.  
Therefore:

**log(n!) = O(n·log(n))**

Shuffling is the process of randomly rearranging the elements in a list or sequence. It ensures that each possible arrangement has an equal chance of occurring. Shuffling is widely used in gaming (e.g., shuffling cards), machine learning (e.g., shuffling training data), and simulations.

## Fisher-Yates Shuffle Algorithm

## What is it?

The **Fisher-Yates Shuffle** is an algorithm for generating a random permutation of a finite list — in other words, for **shuffling** a list such that every possible order is equally likely.

It is the **underlying algorithm** used by Python’s `random.shuffle()`.

---

## How it Works (Step-by-Step)

Given a list of `n` elements:

1. Start from the last element (index `n-1`)
2. Pick a random index `j` such that `0 ≤ j ≤ i`
3. Swap element at index `i` with element at index `j`
4. Repeat for `i = n-1` down to `1`

This ensures all permutations occur with equal probability — **O(n)** time and **O(1)** space.

---

## Python Implementation

```python
import random

def fisher_yates_shuffle(arr):
    n = len(arr)
    for i in range(n - 1, 0, -1):
        j = random.randint(0, i)
        arr[i], arr[j] = arr[j], arr[i]

# Example usage
items = [1, 2, 3, 4, 5]
print("Before shuffle:", items)
fisher_yates_shuffle(items)
print("After shuffle :", items)
```

---

## Example Walkthrough

Initial list: `[1, 2, 3, 4, 5]`

Suppose `random.randint()` gives:
- i = 4, j = 2 → swap 5 and 3 → `[1, 2, 5, 4, 3]`
- i = 3, j = 1 → swap 4 and 2 → `[1, 4, 5, 2, 3]`
- i = 2, j = 0 → swap 5 and 1 → `[5, 4, 1, 2, 3]`
- i = 1, j = 1 → no change

Final shuffled list: `[5, 4, 1, 2, 3]` (your result will vary unless seeded)

---

## Why is Fisher-Yates Important?

- **Uniform randomness**: All permutations are equally likely
- **Efficient**: Runs in O(n) time and uses no extra space
- **Used in production**: Behind `random.shuffle()` and similar methods

---

## When to Use

- Shuffling decks/cards
- Randomizing datasets
- Sampling without replacement

# Quick Find

## What is Quick Find?

Quick Find is an algorithm used to solve the **dynamic connectivity problem**, where we determine if two elements are in the same component (i.e., connected). Each element belongs to a group, and groups can be merged.

---

## How it Works

- We maintain an array `id[]`, where `id[i]` is the component id of element `i`.
- `find(p)` returns the id of `p`
- `connected(p, q)` checks if `id[p] == id[q]`
- `union(p, q)` changes all entries with `id[p]` to `id[q]`

---

## Time Complexity

| Operation | Time Complexity |
|-----------|-----------------|
| Find      | O(1)            |
| Union     | O(n)            |

Union is slow because it may scan the entire array.

---

## Example with Custom Elements

Let’s say our elements are: `['a', 'b', 'c', 'x', 'y', 'z']`

We map these to internal indices:
```
Index:   0  1  2  3  4  5
Element: a  b  c  x  y  z
id:      0  1  2  3  4  5
```

Perform `union('a', 'x')`:  
- Internally, `union(0, 3)` → update all `id[i] == id[0]` to `id[3]`

New id array:
```
id:      3  1  2  3  4  5
```

Now `connected('a', 'x')` → True  
But `connected('a', 'y')` → False

---

## Summary

- Good for **static data** with few unions and many `find()` operations
- Poor for **dynamic systems** due to expensive unions


````python
class QuickFindUF:
    def __init__(self, elements):
        self.elements = elements
        self.index = {el: i for i, el in enumerate(elements)}
        self.id = list(range(len(elements)))

    def find(self, p):
        return self.id[self.index[p]]

    def connected(self, p, q):
        return self.find(p) == self.find(q)

    def union(self, p, q):
        pid = self.find(p)
        qid = self.find(q)

        if pid == qid:
            return

        for i in range(len(self.id)):
            if self.id[i] == pid:
                self.id[i] = qid

    def display(self):
        print("id mapping:")
        for el in self.elements:
            print(f"{el}: {self.id[self.index[el]]}", end=" | ")
        print("\n")

# Example
elements = ['a', 'b', 'c', 'x', 'y', 'z']
uf = QuickFindUF(elements)

print("Initial state:")
uf.display()

uf.union('a', 'x')
print("After union('a', 'x'):")
uf.display()

uf.union('b', 'y')
print("After union('b', 'y'):")
uf.display()

print("connected('a', 'x'):", uf.connected('a', 'x'))
print("connected('a', 'y'):", uf.connected('a', 'y'))
````

# Static vs Dynamic Queues

## Static Queue (Array-based)
A static queue uses a fixed-size array. Once full, no more elements can be added unless space is freed (by dequeueing). It is simple but inflexible in memory use.

**Example:** Circular queue in a fixed-size array

---

## Dynamic Queue (Linked List-based)
A dynamic queue uses linked nodes and can grow or shrink at runtime. It is flexible and memory-efficient, as there is no fixed limit.

**Example:** Queue implemented using a linked list


````python
# --------- Static Queue using Array (Fixed Size) ---------
class StaticQueue:
    def __init__(self, capacity):
        self.queue = [None] * capacity
        self.front = 0
        self.rear = -1
        self.size = 0
        self.capacity = capacity

    def enqueue(self, value):
        if self.size == self.capacity:
            print("Queue is full")
            return
        self.rear = (self.rear + 1) % self.capacity
        self.queue[self.rear] = value
        self.size += 1

    def dequeue(self):
        if self.size == 0:
            print("Queue is empty")
            return
        value = self.queue[self.front]
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return value

    def display(self):
        result = []
        for i in range(self.size):
            index = (self.front + i) % self.capacity
            result.append(self.queue[index])
        print("Static Queue:", result)


print("STATIC QUEUE:")
sq = StaticQueue(3)
sq.enqueue(10)
sq.enqueue(20)
sq.enqueue(30)
sq.enqueue(40)  # should print 'Queue is full'
sq.display()
sq.dequeue()
sq.enqueue(40)
sq.display()
````

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class DynamicQueue:
    def __init__(self):
        self.front = self.rear = None

    def enqueue(self, value):
        new_node = Node(value)
        if not self.rear:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

    def dequeue(self):
        if not self.front:
            print("Queue is empty")
            return
        value = self.front.value
        self.front = self.front.next
        if not self.front:
            self.rear = None
        return value

    def display(self):
        result = []
        temp = self.front
        while temp:
            result.append(temp.value)
            temp = temp.next
        print("Dynamic Queue:", result)
print("\nDYNAMIC QUEUE:")
dq = DynamicQueue()
dq.enqueue(1)
dq.enqueue(2)
dq.enqueue(3)
dq.display()
dq.dequeue()
dq.display()
````

## Tree Traversals

Tree traversal is the process of visiting all nodes of a tree in a specific order. There are three main depth-first traversal types for binary trees:

---

## 1. Inorder Traversal (Left, Root, Right)

### Description:
- Visit left subtree
- Visit root node
- Visit right subtree

### Use Case:
- Retrieves nodes in ascending order (for BSTs)

---

## 2. Preorder Traversal (Root, Left, Right)

### Description:
- Visit root node
- Visit left subtree
- Visit right subtree

### Use Case:
- Used to copy or serialize the tree

---

## 3. Postorder Traversal (Left, Right, Root)

### Description:
- Visit left subtree
- Visit right subtree
- Visit root node

### Use Case:
- Used to delete the tree (children before parent)

---

## 4. Level Order Traversal (Breadth-First)

### Description:
- Visit nodes level by level from top to bottom

### Use Case:
- Used in shortest path algorithms or printing trees by level


````python
# Binary Tree Node
class Node:
    def __init__(self, key):
        self.left = None
        self.right = None
        self.val = key

# Inorder Traversal (Left, Root, Right)
def inorder(root):
    if root:
        inorder(root.left)
        print(root.val, end=' ')
        inorder(root.right)

# Preorder Traversal (Root, Left, Right)
def preorder(root):
    if root:
        print(root.val, end=' ')
        preorder(root.left)
        preorder(root.right)

# Postorder Traversal (Left, Right, Root)
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.val, end=' ')

# Level Order Traversal (BFS)
from collections import deque
def level_order(root):
    if not root:
        return
    queue = deque([root])
    while queue:
        node = queue.popleft()
        print(node.val, end=' ')
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)

# Create sample binary tree:
#         1
#       /   \
#      2     3
#     / \   /
#    4   5 6

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)

# Perform traversals
print("Inorder Traversal:")
inorder(root)  # Output: 4 2 5 1 6 3

print("\nPreorder Traversal:")
preorder(root)  # Output: 1 2 4 5 3 6

print("\nPostorder Traversal:")
postorder(root)  # Output: 4 5 2 6 3 1

print("\nLevel Order Traversal:")
level_order(root)  # Output: 1 2 3 4 5 6
````

## Red-Black Tree vs B-Tree

| Aspect        | Red-Black Tree                              | B-Tree                                      |
|---------------|----------------------------------------------|---------------------------------------------|
| Structure     | Binary tree (each node has ≤ 2 children)     | Generalized m-way tree (nodes can have many children) |
| Height        | O(log n), but taller than B-Tree             | O(log n), shorter due to wider branching     |
| Use Case      | In-memory structures (e.g., Java TreeMap)     | Disk-based structures (e.g., databases, file systems) |
| Balance       | Maintained using color rules and rotations   | Maintained using node splitting/merging     |
| Insert Time   | O(log n), may need rotations                 | O(log n), may need node splits              |
| Delete Time   | O(log n), with rebalancing (rotations/colors)| O(log n), may need merging or redistribution|
| Search Time   | O(log n)                                     | O(log n)                                    |
| Storage       | Stores only 1 key per node                   | Stores multiple keys per node               |
| Rotations     | Yes (Left/Right)                             | No (uses split and merge)                   |
| Typical Use   | Compiler trees, language runtimes            | Databases, large index systems              |

---

## Example

Let’s insert keys: [10, 20, 30]

### Red-Black Tree
1. Insert 10 → root becomes black  
2. Insert 20 → red child  
3. Insert 30 → triggers rotation and recoloring

Resulting structure:
```
     20B
    /   \
  10R   30R
```

### B-Tree (Order 3)
1. Insert 10 → fits in root  
2. Insert 20 → fits in same node  
3. Insert 30 → node overflows → split into two nodes  
   - Middle key (20) is promoted

Resulting structure:
```
     [20]
    /    \
 [10]   [30]
```

### Insert
```text
Function Insert(root, key):
    1. If root is full, split it and create a new root
    2. Call InsertNonFull(root, key)

Function InsertNonFull(node, key):
    1. If node is a leaf:
        - Insert key into correct position
    2. Else:
        - Find correct child
        - If child is full, split it
        - Recurse on correct child

```text
Function Delete(root, key):
    1. If key is in a leaf node:
        - Remove the key
    2. If key is in an internal node:
        - Replace with predecessor or successor
        - Recurse to delete the replaced key
    3. If child node has fewer keys than allowed:
        - Borrow from sibling or merge children
        - Recurse on proper child
```

Function Search(node, key):
    1. Find the smallest index i such that key ≤ node.key[i]
    2. If key == node.key[i], return True
    3. If node is leaf, return False
    4. Else, recurse on child[i]
```

# Binary Tree – Theory

## What is a Binary Tree?

A binary tree is a hierarchical, non-linear data structure where each node has at most two children: the left and the right child. It is widely used in searching, sorting, expression evaluation, decision-making, and file system representations.

Each node in a binary tree contains:
- A **data value**
- A **reference to the left child**
- A **reference to the right child**

---

## Common Types of Binary Trees

- **Full Binary Tree**: Every node has either 0 or 2 children.
- **Complete Binary Tree**: All levels are completely filled except possibly the last, which is filled from left to right.
- **Perfect Binary Tree**: All internal nodes have two children, and all leaves are at the same level.
- **Binary Search Tree (BST)**: A binary tree in which:
  - The left subtree of a node contains only nodes with values less than the node’s value.
  - The right subtree contains only nodes with values greater than the node’s value.

---

## Applications of Binary Trees

- Representing hierarchical data (e.g., file systems, family trees)
- Efficient searching and sorting (Binary Search Trees)
- Expression evaluation (Expression Trees)
- Priority queues (Heaps)
- Decision-making (Decision Trees)

---

## Summary

Binary trees provide a powerful and flexible structure for managing hierarchical and sorted data. When balanced, they offer efficient operations for insertion, deletion, and searching, often in O(log n) time. They serve as the basis for more advanced structures like AVL trees, Red-Black trees, and heaps.


````python
# Node definition
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Insert node (BST style)
def insert(root, value):
    if root is None:
        return Node(value)
    if value < root.data:
        root.left = insert(root.left, value)
    else:
        root.right = insert(root.right, value)
    return root

# Inorder traversal (Left, Root, Right)
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)

# Preorder traversal (Root, Left, Right)
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)

# Postorder traversal (Left, Right, Root)
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")

# Search in BST
def search(root, key):
    if root is None or root.data == key:
        return root
    if key < root.data:
        return search(root.left, key)
    else:
        return search(root.right, key)

# Find minimum value node in a subtree
def find_min(root):
    current = root
    while current.left:
        current = current.left
    return current

# Delete a node from BST
def delete(root, key):
    if root is None:
        return root

    if key < root.data:
        root.left = delete(root.left, key)
    elif key > root.data:
        root.right = delete(root.right, key)
    else:
        # Node with only one child or no child
        if root.left is None:
            return root.right
        elif root.right is None:
            return root.left
        # Node with two children: get inorder successor (smallest in the right subtree)
        temp = find_min(root.right)
        root.data = temp.data
        root.right = delete(root.right, temp.data)

    return root

# Example usage
if __name__ == "__main__":
    root = None
    for val in [50, 30, 70, 20, 40, 60, 80]:
        root = insert(root, val)

    print("Inorder Traversal:")
    inorder(root)

    print("\nPreorder Traversal:")
    preorder(root)

    print("\nPostorder Traversal:")
    postorder(root)

    print("\n\nSearch for 60:")
    result = search(root, 60)
    print("Found" if result else "Not found")

    print("\nDeleting 70...")
    root = delete(root, 70)

    print("Inorder after deletion:")
    inorder(root)

````

# Hash Table – Full Theory

## What is a Hash Table?

A **Hash Table** is a data structure that stores key-value pairs. It allows for **fast insertion, deletion, and lookup** operations in average-case constant time: **O(1)**.

Hash tables are widely used for:
- Dictionaries/maps
- Caching
- Symbol tables in compilers
- Associative arrays

---

## How Does a Hash Table Work?

A hash table works by:

1. **Using a hash function** to convert a key into an integer index
2. **Storing the value** at that index in an array

### Example:
If the key is `"apple"` and the array has size 10, a hash function might convert `"apple"` to index 3.  
So, `"apple": 50` gets stored at `table[3]`.

---

## Components

### 1. Hash Function
- Converts the key to an index
- Should be fast and distribute keys uniformly
- Example: `hash("key") % size`

### 2. Array (Buckets)
- Stores actual data
- The size of the array determines the number of possible slots

---

## What is a Collision?

A **collision** occurs when two keys hash to the same index.

Example:
- `"cat"` and `"dog"` both hash to index 5

Collisions must be handled using one of the following techniques:

---

## Collision Handling Techniques

### 1. **Chaining**
- Each index holds a **linked list** of key-value pairs
- Colliding elements are stored in the list at that index

### 2. **Open Addressing**
- If a collision occurs, find the next available slot
- Methods include:
  - **Linear Probing**: try next index
  - **Quadratic Probing**: use a quadratic step (i²)
  - **Double Hashing**: use a second hash function

---

## Time Complexities (Average Case)

| Operation | Time Complexity |
|-----------|-----------------|
| Insert    | O(1)            |
| Search    | O(1)            |
| Delete    | O(1)            |

In the **worst case** (due to many collisions), operations may degrade to **O(n)**.

---

## Load Factor

- Defined as: `load_factor = number_of_elements / table_size`
- A high load factor increases collision risk
- To maintain performance, hash tables are resized when load factor exceeds a threshold (e.g., 0.75)

---

## Advantages

- Fast access, insertion, deletion (on average)
- Efficient for large datasets
- Easy to implement key-value mappings

---

## Disadvantages

- Poor performance with bad hash functions or high collision rates
- Not good for ordered data traversal
- Space may be wasted if array is sparsely filled

---

## Real-World Usage

- Python’s `dict` and `set` use hash tables internally
- Java’s `HashMap`, `Hashtable`
- C++’s `unordered_map`
- C’s `hashmap` implementations in libraries

---

## Summary

A hash table is a powerful and efficient data structure used to store key-value pairs. It relies on a hash function to compute an index, allowing average-case constant-time operations. Handling collisions and maintaining a good load factor are key to ensuring performance.


````python
# Hash Table Implementation with Open Addressing (Linear Probing)

class HashTable:
    def __init__(self, size=10):
        self.size = size
        self.keys = [None] * size
        self.values = [None] * size

    def _hash(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        idx = self._hash(key)
        start_idx = idx
        while self.keys[idx] is not None and self.keys[idx] != key:
            idx = (idx + 1) % self.size
            if idx == start_idx:
                raise Exception("Hash Table is full")
        self.keys[idx] = key
        self.values[idx] = value

    def search(self, key):
        idx = self._hash(key)
        start_idx = idx
        while self.keys[idx] is not None:
            if self.keys[idx] == key:
                return self.values[idx]
            idx = (idx + 1) % self.size
            if idx == start_idx:
                break
        return None

    def delete(self, key):
        idx = self._hash(key)
        start_idx = idx
        while self.keys[idx] is not None:
            if self.keys[idx] == key:
                self.keys[idx] = None
                self.values[idx] = None
                self._rehash_from(idx)
                return True
            idx = (idx + 1) % self.size
            if idx == start_idx:
                break
        return False

    def _rehash_from(self, delete_idx):
        idx = (delete_idx + 1) % self.size
        while self.keys[idx] is not None:
            key_to_rehash = self.keys[idx]
            val_to_rehash = self.values[idx]
            self.keys[idx] = None
            self.values[idx] = None
            self.insert(key_to_rehash, val_to_rehash)
            idx = (idx + 1) % self.size

    def display(self):
        for i in range(self.size):
            print(f"Index {i}: {self.keys[i]} -> {self.values[i]}")
````

````python
# Create hash table
ht = HashTable()

# Insert key-value pairs
ht.insert("apple", 10)
ht.insert("banana", 20)
ht.insert("orange", 30)
ht.insert("grape", 40)

print("\nHash Table after insertions:")
ht.display()

# Search for keys
print("\nSearch Results:")
print("apple:", ht.search("apple"))
print("banana:", ht.search("banana"))
print("mango:", ht.search("mango"))  # not present

# Delete a key
ht.delete("banana")
print("\nHash Table after deleting 'banana':")
ht.display()

# Insert a new key
ht.insert("mango", 25)
print("\nHash Table after inserting 'mango':")
ht.display()
````

# Time and Space Complexity

Time and space complexity are key concepts used to analyze how efficient an algorithm is in terms of:

- **Time Complexity** → How long the algorithm takes to run
- **Space Complexity** → How much extra memory it uses

They are typically expressed in **Big-O notation**, which describes the **upper bound** of growth relative to the input size `n`.

---

## 1. Time Complexity

Time complexity tells you how the number of steps in an algorithm increases as the input size increases.

---

### a. Nested Loops

Nested loops mean loops within loops. The total number of operations is **multiplicative** based on the levels of nesting.

#### Example:
```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

#### Explanation:

- The **outer loop** runs `n` times.
- For **each** iteration of the outer loop, the **inner loop** runs `n` times.
- So the total operations = `n × n` = `n²`.

If `n = 1000`, this will perform **1,000,000 operations**.

#### Time Complexity: `O(n²)`

This is called **quadratic time** and is typical for comparing all pairs or checking combinations.

---

### b. Recursive Loops

Recursion is when a function **calls itself**. The time complexity of recursion depends on:

1. **How many recursive calls are made per level**
2. **How deep the recursion goes**

#### Simple Example – Linear Recursion:
```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

Each call to `factorial(n)` results in one call to `factorial(n-1)`, until `n == 0`.

- For `n = 5`, the calls look like:
  `factorial(5)` → `factorial(4)` → `factorial(3)` → ... → `factorial(0)`

#### Time Complexity: `O(n)`
#### Space Complexity: `O(n)`

Each call uses stack memory, so for `n` recursive calls, you use `O(n)` space.

#### More Complex Recursion – Tree Recursion:

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)
```

Each call makes **two** more calls. This forms an exponential number of calls.

#### Time Complexity: `O(2ⁿ)`
#### Space Complexity: `O(n)` (due to call stack depth)

---

### c. While Loops

While loops keep running until a condition is false. The time depends on how **fast** the condition becomes false.

#### Example 1 – Linear:
```python
i = 0
while i < n:
    i += 1
```
Runs `n` times.

#### Time Complexity: `O(n)`

---

#### Example 2 – Logarithmic:
```python
i = 1
while i < n:
    i *= 2
```

Here, `i` doubles every time:  
1 → 2 → 4 → 8 → ... → n

You double `i` until it reaches or exceeds `n`. So how many times can you double before reaching `n`?

- This is the same as solving:  
  `2^k = n` ⇒ `k = log₂(n)`

#### Time Complexity: `O(log n)`

This is very efficient for large values of `n`.

---

## 2. Space Complexity

Space complexity refers to the **extra memory** your algorithm needs **aside from the input**.

This includes:

- Temporary variables
- Recursion stack
- Data structures like arrays, hash maps, etc.

---

### Examples:

#### Example 1 – No extra space:
```python
def sum_list(arr):
    total = 0
    for num in arr:
        total += num
    return total
```

Uses just one variable → `O(1)` space.

---

#### Example 2 – Extra array:
```python
def double_elements(arr):
    result = []
    for num in arr:
        result.append(2 * num)
    return result
```

Creates a new array of size `n` → `O(n)` space.

---

## 3. How to Analyze a Given Question

To compute time and space complexity:

1. **Count loops** – Are they nested? Is there recursion?
2. **Track growth** – Are you adding elements to an array/list/map?
3. **Ignore constants** – We use Big-O to focus on growth, not small optimizations
4. **Focus on worst-case** – Big-O reflects the worst-case performance

---

## Summary Table

| Code Pattern                        | Time Complexity | Space Complexity |
|------------------------------------|------------------|------------------|
| Single `for` loop                  | O(n)             | O(1)             |
| Nested `for` loops                 | O(n²)            | O(1)             |
| Recursion (1 call per level)       | O(n)             | O(n)             |
| Recursion (2 calls per level)      | O(2ⁿ)            | O(n)             |
| While loop doubling (`i *= 2`)     | O(log n)         | O(1)             |
| Using extra array of size `n`      | O(n)             | O(n)             |

# What is the Minimum Number of Stacks Required to Implement a Queue?

## Introduction

A **queue** is a First-In-First-Out (FIFO) data structure, where:
- Elements are added at the **rear** (enqueue)
- Elements are removed from the **front** (dequeue)

A **stack**, on the other hand, is a Last-In-First-Out (LIFO) structure, where:
- Elements are both added and removed from the **top**

---

## The Problem

Since stacks and queues have **opposite order behaviors**, the question is:

> How many stacks are needed, at minimum, to simulate a queue?

---

## Answer

###  Minimum Number of Stacks Required: **Two**

With **two stacks**, we can successfully implement a queue that supports both `enqueue()` and `dequeue()` operations.

---

## How It Works

Let’s call the two stacks:

- `stack1` — used for enqueue operations
- `stack2` — used for dequeue operations

### Enqueue Operation:
```text
Push the element onto stack1.
```

### Dequeue Operation:
```text
If stack2 is empty:
    While stack1 is not empty:
        Pop from stack1 and push onto stack2
Pop and return the top element from stack2
```

This reverses the order, thus simulating FIFO behavior.

---

## Step-by-Step Example

### Input sequence:
```text
enqueue(10), enqueue(20), enqueue(30), dequeue()
```

### Execution:

1. `enqueue(10)` → push to stack1 → stack1 = [10]
2. `enqueue(20)` → push to stack1 → stack1 = [10, 20]
3. `enqueue(30)` → push to stack1 → stack1 = [10, 20, 30]
4. `dequeue()`:
    - stack2 is empty, so:
        - Pop 30 from stack1, push to stack2 → stack2 = [30]
        - Pop 20 → stack2 = [30, 20]
        - Pop 10 → stack2 = [30, 20, 10]
    - Now, pop 10 from stack2 → **dequeued 10**

Thus, queue behavior is successfully simulated using two stacks.

---

## Time Complexity

| Operation | Worst Case Time |
|-----------|------------------|
| Enqueue   | O(1)             |
| Dequeue   | O(n) in worst case, but amortized O(1) over time |

---

## Can We Do It With One Stack?

**No**, a single stack cannot simulate queue behavior directly, because:
- There’s no way to access the bottom-most element in a stack efficiently
- Reversing order requires at least one more auxiliary structure



````python
class QueueUsingStacks:
    def __init__(self):
        self.stack1 = []  # Used for enqueue
        self.stack2 = []  # Used for dequeue

    def enqueue(self, value):
        # Always push to stack1
        self.stack1.append(value)
        print(f"Enqueued: {value}")

    def dequeue(self):
        if not self.stack2:
            # Transfer all items from stack1 to stack2 if stack2 is empty
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        if not self.stack2:
            print("Queue is empty")
            return None
        return self.stack2.pop()

    def peek(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        if not self.stack2:
            return None
        return self.stack2[-1]

    def is_empty(self):
        return not self.stack1 and not self.stack2

    def display(self):
        # Show contents as a queue from front to rear
        full = self.stack2[::-1] + self.stack1
        print("Queue:", full)

# ----------- Example Usage -----------

q = QueueUsingStacks()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

q.display()  # Should show: [10, 20, 30]

print("Dequeued:", q.dequeue())  # Should return 10
q.display()  # Should show: [20, 30]

q.enqueue(40)
q.display()  # Should show: [20, 30, 40]

print("Peek:", q.peek())  # Should return 20
print("Is empty?", q.is_empty())  # Should return False

# Remove all
print("Dequeued:", q.dequeue())
print("Dequeued:", q.dequeue())
print("Dequeued:", q.dequeue())
print("Dequeued:", q.dequeue())  # Should print "Queue is empty"
````

# Merge Sort – Theory and Example

## What is Merge Sort?

Merge Sort is a **divide-and-conquer algorithm** that sorts an array by:

1. Dividing it into two halves
2. Sorting each half recursively
3. Merging the sorted halves into one sorted array

---

## Key Points

- **Time Complexity:** O(n log n) in all cases (best, average, worst)
- **Space Complexity:** O(n)
- **Stable Sort:** Yes
- **Recursive:** Yes
- **Efficient for:** Large datasets, linked lists, external sorting

---

## How Merge Sort Works

### Step-by-Step Example

Let’s sort the list: `[8, 4, 5, 2, 9, 1]`

### Step 1: Divide
Split into halves:
- Left: `[8, 4, 5]`
- Right: `[2, 9, 1]`

Keep dividing until single-element lists:
```
[8], [4], [5], [2], [9], [1]
```

### Step 2: Merge (Sort while merging)
- Merge [8] and [4] → `[4, 8]`
- Merge [4, 8] and [5] → `[4, 5, 8]`
- Merge [2] and [9] → `[2, 9]`
- Merge [2, 9] and [1] → `[1, 2, 9]`

### Step 3: Final Merge
- Merge `[4, 5, 8]` and `[1, 2, 9]` → `[1, 2, 4, 5, 8, 9]`

---

## Why Use Merge Sort?

- Always guarantees O(n log n) time
- Great for linked lists (no random access)
- Performs well with large files (external sort)
- Doesn’t degrade with bad input like quicksort can


````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    # Split
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0

    # Merge
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Example usage
arr = [8, 4, 5, 2, 9, 1]
print("Original:", arr)
sorted_arr = merge_sort(arr)
print("Sorted:  ", sorted_arr)
````

# Derivation: Relationship Between Maximum Nodes and Height of a Perfect Binary Tree

## What is a Perfect Binary Tree?

A **perfect binary tree** is a binary tree in which:

- All **internal nodes** have **two children**
- All **leaves** are at the **same level**
- The tree is **completely filled** at all levels

Let:
- `h` = height of the tree (number of edges from root to deepest leaf)
- `N` = total number of nodes in the tree

---

## Derivation of Maximum Number of Nodes

### Step 1: Observe node count at each level

In a perfect binary tree:
- Level 0 has `1` node
- Level 1 has `2` nodes
- Level 2 has `4` nodes
- ...
- Level `k` has `2^k` nodes

So, total number of nodes `N` is the **sum of a geometric series**:

```
N = 2^0 + 2^1 + 2^2 + ... + 2^h
```

This is a geometric progression with:
- First term `a = 1`
- Ratio `r = 2`
- Number of terms = `h + 1`

### Step 2: Use the formula for sum of GP

```
Sum = a * (r^(n) - 1) / (r - 1)
```

Apply it:
```
N = 1 * (2^(h + 1) - 1) / (2 - 1)
  = 2^(h + 1) - 1
```

---

## Final Relationship

```
N = 2^(h + 1) - 1
```

Where:
- `N` is the maximum number of nodes
- `h` is the height of the tree

---

## Deriving Height from Nodes

To find `h` from a given number of nodes `N`:

```
N = 2^(h + 1) - 1
=> 2^(h + 1) = N + 1
=> h + 1 = log₂(N + 1)
=> h = log₂(N + 1) - 1
```

---

## Examples

| Height (h) | Maximum Nodes (N)        |
|------------|--------------------------|
| 0          | 2^1 - 1 = 1              |
| 1          | 2^2 - 1 = 3              |
| 2          | 2^3 - 1 = 7              |
| 3          | 2^4 - 1 = 15             |
| 4          | 2^5 - 1 = 31             |
| h          | 2^(h + 1) - 1            |

---

## Conclusion

The number of nodes in a perfect binary tree grows **exponentially** with the height:

- If height increases by 1, the number of nodes nearly **doubles**
- This makes perfect binary trees **space-efficient** and **fast** for operations like search, insert, and traversal (all O(log n))

This derivation is essential when analyzing tree performance and estimating memory or node requirements.

### Real-World Example:
Suppose you're looking up a customer's name by their ID:
```sql
SELECT name FROM customers WHERE customer_id = 105;
```

- The database uses a **symbol table-like index** to locate row `105` directly
- Without this, it would scan every row (O(n)); with the index, it’s O(1) or O(log n)

---

## Comparison Table: Symbol Tables in Dictionaries vs Indexing

| Aspect                | Symbol Table in Dictionary (Code)       | Symbol Table in Indexing (Database)     |
|------------------------|------------------------------------------|------------------------------------------|
| Purpose                | Tracks program symbols                   | Maps keys to data for fast lookup         |
| Structure              | Hash table or BST                        | Hash index, B+ Tree, or bitmap index      |
| Key                    | Variable name, function, identifier      | ID, code, label, key                      |
| Value                  | Metadata (type, scope, address, value)   | Data row reference or actual value        |
| Use Case               | Compilers, interpreters, language runtimes | SQL queries, table joins, relational links|
| Scope awareness        | Yes (global, local, block scopes)        | No (flat, key-to-value mapping only)      |
| Lookup Speed           | O(1) average with hashing                | O(1) or O(log n) depending on index type  |
| Mutability             | Supports redefinition                    | Usually read-optimized (less write-friendly) |

---

## Summary

- A **symbol table** is a mapping structure: it relates symbols/keys to values
- In **programming**, symbol tables manage variables, functions, and scopes
- In **indexing**, symbol tables act as lookups to avoid full data scans
- Both use **efficient data structures** (hash tables, trees) for speed and scalability

Understanding both contexts helps in mastering **compilers**, **interpreters**, and **database systems** alike.
