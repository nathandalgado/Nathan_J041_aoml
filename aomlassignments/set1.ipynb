### Singly Linked List – Overview

A **singly linked list** is a linear data structure where each element, called a **node**, contains:

- **Data** – the actual value stored.
- **Next pointer** – a reference to the next node in the sequence.

Unlike arrays, elements in a linked list are **not stored in contiguous memory**. This allows efficient insertion and deletion, especially at the beginning or middle, without needing to shift elements.

### Key Characteristics:
- Begins with a **`head`** node.
- Each node points to the next; the **last node points to `None`**.
- Supports **only forward traversal**.

### Common Operations:
- `append(data)` – Add a node at the end.
- `insert_first(data)` – Add a node at the beginning.
- `insert_at(index, data)` – Insert a node at a specific position.
- `delete_first()` – Remove the first node.
- `delete_last()` – Remove the last node.
- `delete(data)` – Remove the first node with the specified data.
- `display()` – Print all nodes in order.

Singly linked lists are often used when frequent insertions and deletions are needed and random access is not a priority.


````python
# Node class
class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None

# Singly Linked List class
class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    # Insert at end
    def append(self, data):
        node = Node(data)
        if self.tail:
            self.tail.next = node
            self.tail = node
        else:
            self.head = self.tail = node
        self.size += 1

    # Insert at beginning
    def insert_first(self, data):
        node = Node(data)
        if not self.head:
            self.head = self.tail = node
        else:
            node.next = self.head
            self.head = node
        self.size += 1

    # Insert at position
    def insert_at(self, index, data):
        if index == 0:
            self.insert_first(data)
            return
        if index >= self.size:
            self.append(data)
            return

        node = Node(data)
        prev = None
        current = self.head
        for _ in range(index):
            prev = current
            current = current.next
        node.next = current
        prev.next = node
        self.size += 1

    # Delete first node
    def delete_first(self):
        if self.head:
            self.head = self.head.next
            if self.head is None:
                self.tail = None
            self.size -= 1

    # Delete last node
    def delete_last(self):
        if not self.head:
            return
        if self.head == self.tail:
            self.head = self.tail = None
        else:
            prev = self.head
            current = self.head.next
            while current.next:
                prev = current
                current = current.next
            prev.next = None
            self.tail = prev
        self.size -= 1

    # Delete by value
    def delete(self, data):
        current = self.head
        prev = None
        while current:
            if current.data == data:
                if current == self.head:
                    self.delete_first()
                elif current == self.tail:
                    self.delete_last()
                else:
                    prev.next = current.next
                    self.size -= 1
                return
            prev = current
            current = current.next

    # Print list
    def display(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
````

````python
ll = SinglyLinkedList()
ll.append("egg")
ll.append("ham")
ll.append("spam")
ll.insert_first("start")
ll.insert_at(2, "middle")
ll.display()            # start -> egg -> middle -> ham -> spam -> None
ll.delete_first()
ll.delete_last()
ll.delete("middle")
ll.display()            # egg -> ham -> None
````

###🧠 Detecting Line Segment Intersections Using Binary Search Tree

Detecting intersections among a set of line segments is a classic problem in computational geometry. A highly efficient solution uses a **Sweep Line Algorithm** combined with a **Binary Search Tree (BST)**.

### 🔄 Concept Overview

- Imagine a vertical line (called the **sweep line**) moving from left to right across the 2D plane.
- As the line sweeps, it encounters **events**, which are the endpoints of the segments.
- These events are processed in order of their **x-coordinates**.

### 🧰 Data Structures Used

1. **Event Queue (Priority Queue or Sorted List)**:
   - Holds all segment endpoints.
   - Sorted by x-coordinate, and for tie-breaking, by y-coordinate.

2. **Status Structure (BST)**:
   - Maintains the set of **currently active segments** — those intersected by the sweep line at its current x-position.
   - Segments in the BST are ordered by the y-coordinate of their intersection with the sweep line.

### 🛠️ How the Algorithm Works

1. **Sort all segment endpoints** by their x-coordinates to form the event queue.
2. **Sweep the line** from left to right:
   - **Left Endpoint of Segment**:
     - Insert the segment into the BST.
     - Check for intersection with its **immediate upper and lower neighbors**.
   - **Right Endpoint of Segment**:
     - Remove the segment from the BST.
     - After deletion, check if the removed segment’s neighbors now intersect with each other.

3. **Intersection Detection**:
   - Intersections are only possible between **neighboring segments** in the BST.
   - So, we only check these local neighbors at each step, not all pairs.

### 📍 Small Example

Given 3 segments:

- Segment **A**: from (1, 1) to (4, 4)
- Segment **B**: from (1, 4) to (4, 1)
- Segment **C**: from (5, 0) to (6, 2)

**Step-by-step:**

1. **Sort Events**:  
   Sorted by x: (1,1) [A start], (1,4) [B start], (4,1) [B end], (4,4) [A end], (5,0), (6,2)

2. **Process (1,1)** → Insert Segment A into BST  
   BST = [A]

3. **Process (1,4)** → Insert Segment B  
   - BST = [A, B] (ordered by y at x = 1)
   - Check neighbors: A and B
   - ✅ **Intersection detected** at (2.5, 2.5)

4. Algorithm can stop or continue (C is far off and doesn't intersect A or B).

### ✅ Advantages

- **Efficient**: Runs in O(n log n + k log n), where `n` is the number of segments and `k` is the number of intersections.
- **Scalable**: Works well for large datasets.
- **Localized checks**: Only neighboring segments in the BST are compared, avoiding brute-force O(n²) checks.

### 📌 Use Case

Used in:
- GIS and mapping systems
- CAD software
- Game development (collision detection)
- Computer graphics

##  Quick Sort – Overview

**Quick Sort** is a highly efficient and widely used **divide-and-conquer** sorting algorithm.  
It works by selecting a **pivot** element from the array and partitioning the other elements into two sub-arrays:
- Elements **less than or equal** to the pivot
- Elements **greater** than the pivot

These sub-arrays are then **recursively sorted**, and the result is a sorted array.

---

###  Key Characteristics

- **Time Complexity**:
  - Best Case: O(n log n)
  - Average Case: O(n log n)
  - Worst Case: O(n²) — occurs when the smallest or largest element is always chosen as the pivot
- **Space Complexity**: O(log n) for recursion stack (in-place sort)
- **Stability**: Not stable (can be modified to be stable)

---

###  Pseudocode for Quick Sort

```plaintext
QUICKSORT(A, low, high)
    if low < high then
        pivotIndex ← PARTITION(A, low, high)
        QUICKSORT(A, low, pivotIndex - 1)
        QUICKSORT(A, pivotIndex + 1, high)

PARTITION(A, low, high)
    pivot ← A[high]               // choose the last element as pivot
    i ← low - 1                   // index of smaller element

    for j ← low to high - 1 do
        if A[j] ≤ pivot then
            i ← i + 1
            swap A[i] and A[j]

    swap A[i + 1] and A[high]     // place pivot at the correct position
    return i + 1                  // return pivot index


````python
def quick_sort(arr, low, high):
    if low < high:
        pivot_index = partition(arr, low, high)
        quick_sort(arr, low, pivot_index - 1)
        quick_sort(arr, pivot_index + 1, high)

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

# Example usage
arr = [10, 7, 8, 9, 1, 5]
quick_sort(arr, 0, len(arr) - 1)
print("Sorted array:", arr)
````

## Kruskal’s Algorithm

Kruskal’s Algorithm is a greedy algorithm used to find the **Minimum Spanning Tree (MST)** of a **connected, undirected, and weighted graph**. The MST is a subset of the edges that connects all the vertices in the graph with the **minimum possible total edge weight**, without forming any cycles.

Kruskal’s algorithm operates by **sorting all the edges** of the graph in ascending order of weight and then **selecting the smallest edge** that **does not form a cycle** when added to the current MST. To efficiently check for cycles, the algorithm uses the **Union-Find (Disjoint Set Union) data structure**.

## Steps of Kruskal’s Algorithm:
1. Sort all the edges of the graph in non-decreasing order of weight.
2. Initialize an empty set for the MST.
3. Initialize a Union-Find data structure to keep track of connected components.
4. For each edge in the sorted list:
   - If the edge connects vertices in **different sets** (i.e., does not form a cycle), include it in the MST and perform a **union** operation.
   - Otherwise, skip the edge.
5. Stop when the MST contains **(V - 1)** edges, where **V** is the number of vertices.

## Time Complexity:
- Sorting edges: O(E log E)
- Union-Find operations (with optimizations): O(α(V)), nearly constant
- Overall: **O(E log E)**, where E is the number of edges

## Pseudocode for Kruskal’s Algorithm:
```plaintext
KRUSKAL(G):
    MST ← empty set
    sort all edges of G by increasing weight
    create disjoint sets for each vertex (Union-Find)
    
    for each edge (u, v) in sorted edges:
        if FIND(u) ≠ FIND(v):
            add (u, v) to MST
            UNION(u, v)
        else:
            skip (u, v)  // cycle detected

    return MST


## Union-Find (Disjoint Set Union) Data Structure

The **Union-Find** data structure is used to manage a collection of **disjoint sets**. It supports two efficient operations which are useful for handling dynamic connectivity in graphs:

### 1. `FIND(x)`

This operation returns the **representative** or **parent** of the set containing element `x`. It is used to determine which subset a particular element belongs to. The operation is optimized using **path compression**, which flattens the structure of the tree to make future `FIND` operations faster.

### 2. `UNION(x, y)`

This operation **merges** the sets containing elements `x` and `y`. If both elements already belong to the same set, the operation does nothing. To keep the trees shallow and operations fast, **union by rank** (or size) is applied — the root of the tree with smaller rank is made a child of the root with higher rank.

---

### Use in Kruskal’s Algorithm

In **Kruskal’s algorithm**, the Union-Find data structure is essential for **cycle detection**. As edges are considered in increasing order of weight:

- The algorithm uses `FIND` to check if the vertices of the edge belong to the **same set** (i.e., are already connected).
- If they are, including that edge would form a **cycle**, so it is skipped.
- If they are in **different sets**, the edge is safe to add to the MST, and `UNION` is called to merge the sets.

This ensures the resulting spanning tree is **acyclic** and **minimally weighted**.

---

### Pseudocode for Union-Find

```plaintext
MAKE-SET(x):
    parent[x] ← x
    rank[x] ← 0

FIND(x):
    if parent[x] ≠ x:
        parent[x] ← FIND(parent[x])  // Path compression
    return parent[x]

UNION(x, y):
    rootX ← FIND(x)
    rootY ← FIND(y)

    if rootX = rootY:
        return  // already in same set (would create a cycle)

    // Union by rank
    if rank[rootX] < rank[rootY]:
        parent[rootX] ← rootY
    else if rank[rootX] > rank[rootY]:
        parent[rootY] ← rootX
    else:
        parent[rootY] ← rootX
        rank[rootX] ← rank[rootX] + 1

### Singly Linked List – Overview

A **singly linked list** is a linear data structure where each element, called a **node**, contains:

- **Data** – the actual value stored.
- **Next pointer** – a reference to the next node in the sequence.

Unlike arrays, elements in a linked list are **not stored in contiguous memory**. This allows efficient insertion and deletion, especially at the beginning or middle, without needing to shift elements.

### Key Characteristics:
- Begins with a **`head`** node.
- Each node points to the next; the **last node points to `None`**.
- Supports **only forward traversal**.

### Common Operations:
- `append(data)` – Add a node at the end.
- `insert_first(data)` – Add a node at the beginning.
- `insert_at(index, data)` – Insert a node at a specific position.
- `delete_first()` – Remove the first node.
- `delete_last()` – Remove the last node.
- `delete(data)` – Remove the first node with the specified data.
- `display()` – Print all nodes in order.

Singly linked lists are often used when frequent insertions and deletions are needed and random access is not a priority.


````python
# Node class
class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None

# Singly Linked List class
class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    # Insert at end
    def append(self, data):
        node = Node(data)
        if self.tail:
            self.tail.next = node
            self.tail = node
        else:
            self.head = self.tail = node
        self.size += 1

    # Insert at beginning
    def insert_first(self, data):
        node = Node(data)
        if not self.head:
            self.head = self.tail = node
        else:
            node.next = self.head
            self.head = node
        self.size += 1

    # Insert at position
    def insert_at(self, index, data):
        if index == 0:
            self.insert_first(data)
            return
        if index >= self.size:
            self.append(data)
            return

        node = Node(data)
        prev = None
        current = self.head
        for _ in range(index):
            prev = current
            current = current.next
        node.next = current
        prev.next = node
        self.size += 1

    # Delete first node
    def delete_first(self):
        if self.head:
            self.head = self.head.next
            if self.head is None:
                self.tail = None
            self.size -= 1

    # Delete last node
    def delete_last(self):
        if not self.head:
            return
        if self.head == self.tail:
            self.head = self.tail = None
        else:
            prev = self.head
            current = self.head.next
            while current.next:
                prev = current
                current = current.next
            prev.next = None
            self.tail = prev
        self.size -= 1

    # Delete by value
    def delete(self, data):
        current = self.head
        prev = None
        while current:
            if current.data == data:
                if current == self.head:
                    self.delete_first()
                elif current == self.tail:
                    self.delete_last()
                else:
                    prev.next = current.next
                    self.size -= 1
                return
            prev = current
            current = current.next

    # Print list
    def display(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
````

# Binary Tree – Advantages and Disadvantages

## What is a Binary Tree?

A **binary tree** is a data structure in which each node has at most two children, referred to as the **left child** and **right child**. Binary trees are widely used in various computational applications such as searching, sorting, hierarchical modeling, and expression evaluation.

---

## Advantages of Binary Trees

### 1. Efficient Searching, Insertion, and Deletion (O(log n) in balanced trees)

In a **balanced binary search tree (BST)**, the height of the tree is kept to a minimum, which ensures that searching, inserting, and deleting nodes takes logarithmic time: **O(log n)**. This efficiency is a major improvement over linear data structures like arrays or linked lists, which have O(n) time complexity for the same operations.

For example, searching for a value in a balanced BST with 1,000 elements may only require ~10 comparisons, while a linear search in an unsorted array may require up to 1,000.

---

### 2. Natural Representation of Hierarchical Data

Binary trees are well-suited for modeling hierarchical relationships. Common examples include:

- File systems (folders and subfolders)
- Organization charts
- Family trees
- Expression trees in compilers

Their tree-like nature allows data to be structured logically and traversed efficiently.

---

### 3. Easy Retrieval of Sorted Data (In BSTs)

In a **binary search tree**, an **inorder traversal** retrieves elements in ascending order. This eliminates the need for additional sorting algorithms, making BSTs a great choice for applications where **ordered data access** is frequently needed (e.g., dictionaries, leaderboards).

---

### 4. Recursive Structure Matches Recursive Algorithms

Binary trees naturally lend themselves to **recursive solutions**. Each subtree is itself a binary tree, so algorithms like traversals (inorder, preorder, postorder), insertion, and search can be implemented using clean, recursive logic.

This makes them intuitive for many divide-and-conquer approaches, such as in **merge sort** trees or **decision trees**.

---

### 5. Memory Efficiency in Sparse Data Structures

Binary trees allocate memory **dynamically**—only when nodes are created. Unlike arrays, where a fixed amount of space may be reserved regardless of usage, binary trees grow as needed, making them efficient for **sparse** datasets or unknown sizes at runtime.

---

## Disadvantages of Binary Trees

### 1. Performance Degrades if Unbalanced (Worst Case O(n))

In the **worst case**, a binary tree can become **skewed** (e.g., all nodes are added to the right), effectively forming a **linked list**. This leads to degraded performance, where operations like search and insert become **linear (O(n))**, defeating the main advantage of a tree.

Balanced variants (e.g., AVL or Red-Black trees) are needed to prevent this.

---

### 2. Complex Node Deletion Process

Deleting nodes from a binary tree—especially nodes with **two children**—requires complex handling:

- Replace the deleted node with its inorder predecessor or successor
- Maintain BST properties during restructuring

This process is more difficult compared to simple removal in arrays or linked lists and increases the chance of introducing bugs if not handled carefully.

---

### 3. Extra Memory Usage for Pointers

Each node in a binary tree must store two references (left and right child). For large trees, especially sparse ones, this pointer overhead can become significant compared to compact data structures like arrays or hash tables.

This overhead becomes even more pronounced in memory-constrained systems.

---

### 4. Difficult to Visualize and Debug

Unlike linear structures, trees—especially unbalanced or deep ones—are hard to visualize and traverse without recursion or stacks. Debugging tree operations like deletion, balancing, or rotations (in self-balancing trees) requires a good understanding of the structure and careful tracing.

---

### 5. Requires Balancing for Consistent Performance

For consistent and reliable performance, binary trees often need to be **self-balanced**. Without balancing, their operation time can degrade significantly.

This requires using more complex trees like:
- **AVL Trees** (auto-balancing after each insertion/deletion)
- **Red-Black Trees** (used in Java TreeMap, Linux kernel)
- **Splay Trees** (frequently accessed nodes move to the top)

These add extra logic, overhead, and implementation complexity.

---

## Summary

Binary trees are versatile and powerful data structures for organizing hierarchical or sorted data. They offer efficient performance and flexible structure, but only when **well-balanced** and properly implemented. Their drawbacks can be mitigated using self-balancing variants, which are commonly used in databases, compilers, and operating systems.

# Binary Tree – Theory

## What is a Binary Tree?

A binary tree is a hierarchical, non-linear data structure where each node has at most two children: the left and the right child. It is widely used in searching, sorting, expression evaluation, decision-making, and file system representations.

Each node in a binary tree contains:

## Common Types of Binary Trees

- **Full Binary Tree**: Every node has either 0 or 2 children.
- **Complete Binary Tree**: All levels are completely filled except possibly the last, which is filled from left to right.
- **Perfect Binary Tree**: All internal nodes have two children, and all leaves are at the same level.
- **Binary Search Tree (BST)**: A binary tree in which:
  - The left subtree of a node contains only nodes with values less than the node’s value.
  - The right subtree contains only nodes with values greater than the node’s value.

---

## Applications of Binary Trees

- Representing hierarchical data (e.g., file systems, family trees)
- Efficient searching and sorting (Binary Search Trees)
- Expression evaluation (Expression Trees)
- Priority queues (Heaps)
- Decision-making (Decision Trees)

---

## Summary

Binary trees provide a powerful and flexible structure for managing hierarchical and sorted data. When balanced, they offer efficient operations for insertion, deletion, and searching, often in O(log n) time. They serve as the basis for more advanced structures like AVL trees, Red-Black trees, and heaps.


````python
# Node definition
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Insert node (BST style)
def insert(root, value):
    if root is None:
        return Node(value)
    if value < root.data:
        root.left = insert(root.left, value)
    else:
        root.right = insert(root.right, value)
    return root

# Inorder traversal (Left, Root, Right)
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)

# Preorder traversal (Root, Left, Right)
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)

# Postorder traversal (Left, Right, Root)
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")

# Search in BST
def search(root, key):
    if root is None or root.data == key:
        return root
    if key < root.data:
        return search(root.left, key)
    else:
        return search(root.right, key)

# Example usage
if __name__ == "__main__":
    root = None
    for val in [50, 30, 70, 20, 40, 60, 80]:
        root = insert(root, val)

    print("Inorder Traversal:")
    inorder(root)
    print("\nPreorder Traversal:")
    preorder(root)
    print("\nPostorder Traversal:")
    postorder(root)

    print("\n\nSearch for 60:")
    result = search(root, 60)
    print("Found" if result else "Not found")
````

````python
# Hash Table Implementation with Open Addressing (Linear Probing)

class HashTable:
    def __init__(self, size=10):
        self.size = size
        self.keys = [None] * size
        self.values = [None] * size

    def _hash(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        idx = self._hash(key)
        start_idx = idx
        while self.keys[idx] is not None and self.keys[idx] != key:
            idx = (idx + 1) % self.size
            if idx == start_idx:
                raise Exception("Hash Table is full")
        self.keys[idx] = key
        self.values[idx] = value

    def search(self, key):
        idx = self._hash(key)
        start_idx = idx
        while self.keys[idx] is not None:
            if self.keys[idx] == key:
                return self.values[idx]
            idx = (idx + 1) % self.size
            if idx == start_idx:
                break
        return None

    def delete(self, key):
        idx = self._hash(key)
        start_idx = idx
        while self.keys[idx] is not None:
            if self.keys[idx] == key:
                self.keys[idx] = None
                self.values[idx] = None
                self._rehash_from(idx)
                return True
            idx = (idx + 1) % self.size
            if idx == start_idx:
                break
        return False

    def _rehash_from(self, delete_idx):
        idx = (delete_idx + 1) % self.size
        while self.keys[idx] is not None:
            key_to_rehash = self.keys[idx]
            val_to_rehash = self.values[idx]
            self.keys[idx] = None
            self.values[idx] = None
            self.insert(key_to_rehash, val_to_rehash)
            idx = (idx + 1) % self.size

    def display(self):
        for i in range(self.size):
            print(f"Index {i}: {self.keys[i]} -> {self.values[i]}")
````

````python
# Create hash table
ht = HashTable()

# Insert key-value pairs
ht.insert("apple", 10)
ht.insert("banana", 20)
ht.insert("orange", 30)
ht.insert("grape", 40)

print("\nHash Table after insertions:")
ht.display()

# Search for keys
print("\nSearch Results:")
print("apple:", ht.search("apple"))
print("banana:", ht.search("banana"))
print("mango:", ht.search("mango"))  # not present

# Delete a key
ht.delete("banana")
print("\nHash Table after deleting 'banana':")
ht.display()

# Insert a new key
ht.insert("mango", 25)
print("\nHash Table after inserting 'mango':")
ht.display()
````

# Big-O Notation Verification

## Problem:
Determine whether:

**F(n) = n!**  
**G(n) = nⁿ**  

Satisfies:  
**F(n) = O(G(n))**

---

## Step 1: Recall the Definition of Big-O

By definition,  
**F(n) = O(G(n))** means:

There exist constants **c > 0** and **n₀ ≥ 0** such that:  
**F(n) ≤ c × G(n)** for all **n ≥ n₀**

So we need to prove that:

**n! ≤ c × nⁿ** for some constant **c**, and for sufficiently large **n**

---

## Step 2: Compare the Growth of n! and nⁿ

We consider the ratio:

**(n! / nⁿ)**

We know that:

- **n! = 1 × 2 × 3 × ... × n**  
- **nⁿ = n × n × n × ... × n (n times)**

So:

**n! / nⁿ = (1/n) × (2/n) × (3/n) × ... × (n/n)**

This is the product of **n fractions**, all of which are **≤ 1**  
(and many are much **less than 1** as n grows).

Hence:

**n! / nⁿ < 1 for all n ≥ 1**  
⇒ **n! < nⁿ**

---

## Step 3: Choose Constants

Since **n! < nⁿ** for all n ≥ 1, we can choose:

- **c = 1**
- **n₀ = 1**

So that:

**n! ≤ c × nⁿ** holds for all **n ≥ n₀**

---

## ✅ Final Conclusion:

The condition for Big-O is satisfied. Therefore:

**F(n) = n! is O(nⁿ)**

This means the factorial function grows **slower** than the exponential power **nⁿ**,  
and Big-O notation accurately captures this upper bound.

# Perfect Binary Tree – Maximum Height and Maximum Number of Nodes

## What is a Perfect Binary Tree?

A **perfect binary tree** is a type of binary tree where:
- All internal nodes have **exactly two children**.
- All leaf nodes are at the **same level**.
- The tree is **completely filled** at all levels.

---

## Maximum Number of Nodes

In a perfect binary tree of height `h`, each level `i` contains `2^i` nodes.

So the total number of nodes `n` is:

n = 2^0 + 2^1 + 2^2 + ... + 2^h  
  = 2^(h + 1) - 1

### Formula:
**Max nodes = 2^(h + 1) - 1**

---

## Maximum Height

If the number of nodes `n` is known, we can find the maximum height `h` using:

n = 2^(h + 1) - 1  
⇒ h = log₂(n + 1) - 1

### Formula:
**Max height = log₂(n + 1) - 1**

> Use `floor()` if you want the largest integer value of `h`.

---

## Example

### Case 1: Given height h = 3  
Max nodes = 2^(3 + 1) - 1 = 16 - 1 = **15 nodes**

### Case 2: Given number of nodes n = 31  
Max height = log₂(31 + 1) - 1 = log₂(32) - 1 = **4**

---

## Summary Table

| Property                 | Formula                          |
|--------------------------|----------------------------------|
| Max Nodes (given height) | 2^(h + 1) - 1                    |
| Max Height (given nodes) | log₂(n + 1) - 1                  |

---

## Applications

Perfect binary trees are used in:
- **Binary Heaps**
- **Tournament Trees**
- **Static Balanced Search Trees**
- Any scenario where consistent, minimal tree height is needed

At height h:
- Number of leaves = 2^h
- Total number of keys n (minimum) = 2^h - 1

So:
**Minimum keys:**

To get **maximum number of keys**, assume the tree is made of only **3-nodes** (each node has 3 children and 2 keys).

- Number of leaves = 3^h

- Total number of keys = 2 × number of internal nodes

n = 2 × ((3^h - 1) / 2) = 3^h - 1  
⇒ h = log₃(n + 1)

# Q 7 Heap Sort – Definition and Properties

## What is Heap Sort?

**Heap Sort** is a popular and efficient **comparison-based sorting algorithm** that uses a **binary heap** data structure to sort elements.

It works by:
1. **Building a max heap** from the input data.
2. Repeatedly **extracting the maximum element** from the heap and placing it at the end of the array.
3. **Restoring the heap property** after each extraction using a procedure called **heapify**.

This process continues until all elements are sorted.

---

## Properties of Heap Sort

| Property           | Description                                               |
|--------------------|-----------------------------------------------------------|
| **Time Complexity**| O(n log n) in all cases (best, average, and worst)        |
| **Space Complexity**| O(1) – Heap Sort is an in-place sorting algorithm        |
| **Stability**      | ❌ Not stable – equal elements may not retain order       |
| **Sorting Order**  | Ascending (using max heap) or descending (using min heap)|
| **Type**           | Comparison-based, non-adaptive, in-place sort             |
| **Data Structure** | Binary Heap (usually implemented as an array)             |

---

## Key Advantages

- Guaranteed O(n log n) performance, even in the worst case.
- No additional memory is needed for sorting.
- Works well in systems where memory usage needs to be minimized.

---

## Limitations

- Not stable (order of equal elements is not preserved).
- More cache misses compared to algorithms like quicksort or mergesort.

##Heap Sort – Algorithm Explanation


## Step-by-Step Heap Sort Algorithm

### Step 1: Build a Max Heap
- Convert the unsorted array into a **max heap**, where the **largest element is at the root** (index 0).
- This is done by calling the **heapify()** function from the last non-leaf node up to the root.
- The last non-leaf node is at index `(n // 2) - 1`.

### Step 2: Swap and Heapify
- Swap the root (maximum value) with the **last element** of the heap.
- Reduce the heap size by 1 (exclude the last element from further heapifying — it's now in correct position).
- Call **heapify()** on the new root to maintain the max heap property.

### Step 3: Repeat
- Continue swapping the root with the last remaining unsorted element, reducing the heap size, and calling heapify until the heap size becomes 1.

---

## Key Function: Heapify

The **heapify()** function ensures the max-heap property is maintained:
- If the current node is smaller than one of its children, swap it with the largest child.
- Recursively heapify the affected subtree.

---

## Time Complexity

| Operation         | Time        |
|------------------|-------------|
| Building Max Heap| O(n)        |
| Heapify          | O(log n)    |
| Total Sort Time  | O(n log n)  |

---

## Example (Array: [4, 10, 3, 5, 1])

1. Build max heap → [10, 5, 3, 4, 1]  
2. Swap 10 and 1 → [1, 5, 3, 4, 10] → Heapify → [5, 4, 3, 1, 10]  
3. Swap 5 and 1 → [1, 4, 3, 5, 10] → Heapify → [4, 1, 3, 5, 10]  
4. Repeat until sorted → Final output: [1, 3, 4, 5, 10]

---

Heap Sort ensures all elements are sorted using the heap structure and without additional memory allocation.


````python
# Function to maintain the max-heap property
def heapify(arr, n, i):
    largest = i        # Initialize largest as root
    left = 2 * i + 1   # Left child index
    right = 2 * i + 2  # Right child index

    # If left child is larger than root
    if left < n and arr[left] > arr[largest]:
        largest = left

    # If right child is larger than largest so far
    if right < n and arr[right] > arr[largest]:
        largest = right

    # If largest is not root, swap and continue heapifying
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # Swap
        heapify(arr, n, largest)  # Recursively heapify the affected subtree

# Main Heap Sort function
def heap_sort(arr):
    n = len(arr)

    # Step 1: Build max heap (start from last non-leaf node and heapify)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Step 2: Extract elements one by one
    for i in range(n - 1, 0, -1):
        # Swap current root with the last element
        arr[0], arr[i] = arr[i], arr[0]
        # Heapify the reduced heap
        heapify(arr, i, 0)

# Example usage
arr = [4, 10, 18, 3, 5, 6, 1]
print("Original array:", arr)

heap_sort(arr)
print("Sorted array:  ", arr)
````
